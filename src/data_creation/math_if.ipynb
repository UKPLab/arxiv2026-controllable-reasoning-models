```python
import json
import glob
import os
```

```python
gsm8k = []
for path in glob.glob("../../data/math_if_original/gsm8k*.jsonl"):
    with open(path, "r") as f:
        for line in f:
            gsm8k.append(json.loads(line))
```

```python
len(gsm8k)
```

```python
gsm8k[-10]
```

```python
final_instruction = "The instructions apply to the reasoning part (between <think> and </think>) and the final answer (after </think>)."
```

```python
for x in gsm8k:
    # rename id to key
    x['key'] = x.pop('id')
    # rename constraint_name to instruction_id_list
    x['instruction_id_list'] = x.pop('constraint_name')
    # rename constraint_args to kwargs
    x['kwargs'] = x.pop('constraint_args')
    # Create instruction
    instruction = " ".join(x['constraint_desc']) + " " + final_instruction
    question = x['question'].replace(" Let\'s think step by step and output the final answer after \"####\".", "")
    x['prompt'] = question + " " + instruction
```

```python
with open("../../data/math_if/dev.jsonl", "w") as f:
    for x in gsm8k:
        f.write(json.dumps(x) + "\n")
```

```python
list_paths = glob.glob("../../data/math_if_original/*.jsonl")
# remove gsm8k files
list_paths = [p for p in list_paths if "gsm8k" not in p]

test_set = []
for path in list_paths:
    with open(path, "r") as f:
        for line in f:
            test_set.append(json.loads(line))

for x in test_set:
    # rename id to key
    x['key'] = x.pop('id')
    # rename constraint_name to instruction_id_list
    x['instruction_id_list'] = x.pop('constraint_name')
    # rename constraint_args to kwargs
    x['kwargs'] = x.pop('constraint_args')
    # Create instruction
    instruction = " ".join(x['constraint_desc']) + " " + final_instruction
    question = x['question'].replace(" Let\'s think step by step and output the final answer after \"####\".", "")
    x['prompt'] = question + " " + instruction

with open("../../data/math_if/test.jsonl", "w") as f:
    for x in test_set:
        f.write(json.dumps(x) + "\n")
```
